# 2.. Financial Text Classification (Multi-Class)

```python
!pip install transformers datasets

from transformers import AutoTokenizer, AutoModelForSequenceClassification, AutoModelForTokenClassification, pipeline
```

# Sample Dataset

```python
texts = [
    "Revenue grew 30% after new product",
    "Company announced 1000 layoffs across divisions",
    "Risks include inflation and regulatory pressure",
    "Acquired a fintech firm for $2B",
    "Forecast revised down for next quarter"
]
labels = ["GROWTH", "LAYOFF", "RISK", "ACQUISITION", "EARNINGS_WARNING"]
```

# Label Mapping

```python
label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}
label_ids = [label2id[l] for l in labels]
```

# Convert to Hugging Face Dataset

```python
from datasets import Dataset

data = {"text": texts, "label": label_ids}
ds = Dataset.from_dict(data)
```

# Tokenize

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("yiyanghkust/finbert-pretrain")
ds = ds.map(lambda x: tokenizer(x["text"], truncation=True, padding=True), batched=True)
```

#  Load Model with Correct Labels

```python
from transformers import AutoModelForSequenceClassification



model = AutoModelForSequenceClassification.from_pretrained(
    "yiyanghkust/finbert-pretrain",
    num_labels=len(label2id),
    id2label=id2label,
    label2id=label2id,
    device_map="auto",
)
```

# Train the Model

```python
from transformers import Trainer, TrainingArguments

training_args = TrainingArguments(
    output_dir="./finbert-events",
    per_device_train_batch_size=2,
    num_train_epochs=2,
    logging_steps=1,
    save_strategy="epoch"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=ds,
    tokenizer=tokenizer
)

trainer.train()
```

# Predict (After Training)

```python
import torch

def classify_financial_text(text, device=torch.device("cuda")):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(device)
    logits = model(**inputs).logits
    pred = logits.argmax(dim=-1).item()
    return model.config.id2label[pred]

print(classify_financial_text("Revenue grew 30% after new product"))  # Should now return "GROWTH"
```

# Event & Risk Detection via Classification



# Use BERT/FinBERT to classify financial text into event types: "LAYOFF", "MERGER", "RISK", etc.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, Trainer, TrainingArguments
from datasets import Dataset

# Sample training data
texts = [
    "Company announced 100 layoffs across divisions.",
    "Apple acquired Beats for $3B.",
    "Inflation pressures continue to threaten margins.",
    "Company merged with a local competitor.",
    "Company expects revenue growth next quarter."
]
labels = ["LAYOFF", "ACQUISITION", "RISK", "MERGER", "GROWTH"]

# Mapping
label2id = {label: i for i, label in enumerate(labels)}
id2label = {i: label for label, i in label2id.items()}
label_ids = [label2id[l] for l in labels]

# Dataset
ds = Dataset.from_dict({"text": texts, "label": label_ids})

# Tokenizer & tokenization
tokenizer = AutoTokenizer.from_pretrained("yiyanghkust/finbert-pretrain")
ds = ds.map(lambda x: tokenizer(x["text"], truncation=True, padding=True), batched=True)
```

# Train the Event Classifier

```python
model = AutoModelForSequenceClassification.from_pretrained(
    "yiyanghkust/finbert-pretrain",
    num_labels=len(label2id),
    label2id=label2id,
    id2label=id2label
)

training_args = TrainingArguments(
    output_dir="./finbert-events",
    per_device_train_batch_size=2,
    num_train_epochs=5,
    logging_steps=1,
    save_strategy="no"
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=ds,
    tokenizer=tokenizer
)

trainer.train()
```

# Inference Function

```python
import torch
def classify_financial_text(text, device=torch.device("cuda")):
    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(device)
    logits = model(**inputs).logits
    pred = logits.argmax(dim=-1).item()
    return model.config.id2label[pred]

print(classify_financial_text("Company announced 100 layoffs across divisions."))  # Output: "LAYOFF"
```

# Relation Extraction (FinRE-style)






Use BERT to classify relations between two entities using the format:
ENTITY_A [SEP] sentence [SEP] ENTITY_B



 Define a label2id and id2label mapping before training:

```python
relation_data = {
    "input_text": [
        "Apple [SEP] Apple acquired Beats in 2014. [SEP] Beats",
        "SoftBank [SEP] SoftBank funded Grab to expand in Asia. [SEP] Grab"
    ],
    "label": ["acquired_by", "funded_by"]
}

rel_label2id = {label: i for i, label in enumerate(set(relation_data["label"]))}
rel_id2label = {i: label for label, i in rel_label2id.items()}
rel_data = {
    "input_text": relation_data["input_text"],
    "label": [rel_label2id[l] for l in relation_data["label"]]
}
rel_ds = Dataset.from_dict(rel_data)

rel_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
rel_ds = rel_ds.map(lambda x: rel_tokenizer(x["input_text"], truncation=True, padding=True), batched=True)

```

# Train the Relation Classifier

```python
from transformers import AutoModelForSequenceClassification

rel_model = AutoModelForSequenceClassification.from_pretrained(
    "bert-base-uncased",
    num_labels=len(rel_label2id),
    label2id=rel_label2id,
    id2label=rel_id2label
)

rel_args = TrainingArguments(
    output_dir="./bert-rel",
    per_device_train_batch_size=2,
    num_train_epochs=5,
    logging_steps=1,
    save_strategy="no"
)

rel_trainer = Trainer(
    model=rel_model,
    args=rel_args,
    train_dataset=rel_ds,
    tokenizer=rel_tokenizer
)

rel_trainer.train()
```

# Inference Function

```python
import torch
def predict_relation(text, device=torch.device("cuda")):
    inputs = rel_tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(device)
    logits = rel_model(**inputs).logits
    pred = logits.argmax(dim=-1).item()
    return rel_model.config.id2label[pred]

print(predict_relation("Apple [SEP] Apple acquired Beats in 2014. [SEP] Beats"))  # Output: "acquired_by"

```

 Inspect All Output Probabilities

```python
import torch
import torch.nn.functional as F

def predict_relation_verbose(text, device=torch.device("cuda")):
    inputs = rel_tokenizer(text, return_tensors="pt", truncation=True, padding=True).to(device)
    logits = rel_model(**inputs).logits
    probs = F.softmax(logits, dim=1).detach().cpu().numpy()[0]
    return {rel_model.config.id2label[i]: round(p, 3) for i, p in enumerate(probs)}

print(predict_relation_verbose("Apple [SEP] Apple acquired Beats in 2014. [SEP] Beats"))
```
